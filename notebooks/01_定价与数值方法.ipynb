# 定价与数值方法
这里的输入是数学练习参数，不是50ETF市场报价。先预测结果方向，再运行。

```python
from etf_option.pricing import price, greeks, parity, implied_vol
S, K, T, r, q, sigma = 100., 105., .5, .03, .01, .2
c = price(S,K,T,r,sigma,q,'C')
p = price(S,K,T,r,sigma,q,'P')
print({'call':c, 'put':p, 'parity_error':c-p-parity(S,K,T,r,q)})
```

## 把价格反过来求波动率
两个算法应该回收到输入sigma。注意这里检验数学，不是检验赚钱。

```python
for method in ['brent','newton']:
    recovered = implied_vol(c,S,K,T,r,q,'C',method)
    print(method, recovered, recovered-sigma)
print(greeks(S,K,T,r,sigma,q,'C'))
```

## 自己检查Delta
把h从1e-2改到1e-8，理解截断误差和浮点误差。

```python
h = 1e-3
fd = (price(S+h,K,T,r,sigma,q)-price(S-h,K,T,r,sigma,q))/(2*h)
print({'finite_difference_delta':fd, 'analytic_delta':greeks(S,K,T,r,sigma,q)['delta']})
```
